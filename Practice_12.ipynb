```python
words = {"a": "apple", "b": "banana", "c": "elephant", "d": "cat"}
count = 0

for word in words.values():
    if len(word) > 5:
        count += 1

print("Count:", count)
```

```python
words = {"a": "apple", "b": "banana", "c": "orange"}

for word in words.values():
    if "a" in word:
        print(word, "contains a")
    else:
        print(word, "does not contain a")
```

```python
stock = {"Pen": 10, "Book": 0, "Pencil": 5}

for item, quantity in stock.items():
    if quantity > 0:
        print(item, "Available")
    else:
        print(item, "Out of stock")
```

```python
prices = {"Shirt": 500, "Shoes": 1500, "Bag": 800}

for item, price in prices.items():
    if price >= 1000:
        discount = price * 0.10
    else:
        discount = price * 0.05

    print(item, "Final price:", price - discount)
```

```python
attendance = {
    "Rahul": "Present",
    "Priya": "Absent",
    "Amit": "Present"
}

for name, status in attendance.items():
    if status == "Present":
        print(name, "Attended school")
    else:
        print(name, "Was absent")
```

```python
balance = {"Rahul": 5000, "Priya": 500, "Amit": 0}

for name, amount in balance.items():
    if amount >= 1000:
        print(name, "Good balance")
    elif amount > 0:
        print(name, "Low balance")
    else:
        print(name, "No balance")
```

```python
products = {"Pen": 10, "Bag": 800, "Laptop": 50000, "Book": 100}

for name, price in products.items():
    if price > 1000:
        print(name, "Expensive")
    else:
        print(name, "Affordable")
```

```python
ages = {"Rahul": 5, "Priya": 8, "Amit": 3}

for name, age in ages.items():
    if age >= 5:
        print(name, "Eligible for admission")
    else:
        print(name, "Not eligible")
```
