```python
text = input("Enter a sentence: ")
words = text.split()

if words:
    shortest = words[0]

    for word in words:
        if len(word) < len(shortest):
            shortest = word

    print("Shortest word:", shortest)
else:
    print("No words entered")
```

```python
text = input("Enter a string: ")
frequency = {}

for ch in text:
    if ch in frequency:
        frequency[ch] += 1
    else:
        frequency[ch] = 1

print(frequency)
```

```python
text = input("Enter a string: ")
result = ""

for ch in text:
    if ch not in result:
        result += ch

print(result)
```

```python
text1 = input("Enter first string: ").lower()
text2 = input("Enter second string: ").lower()

if sorted(text1) == sorted(text2):
    print("Anagrams")
else:
    print("Not anagrams")
```

```python
numbers = {10, 20, 30, 40}
total = 0

for n in numbers:
    total += n

print("Sum:", total)
```

```python
numbers = {10, 20, 30, 40, 50}
count = 0

for n in numbers:
    count += 1

print("Count:", count)
```

```python
numbers = {15, 45, 25, 80, 30}
largest = None

for n in numbers:
    if largest is None or n > largest:
        largest = n

print("Largest:", largest)
```

```python
numbers = {15, 45, 25, 80, 30}
smallest = None

for n in numbers:
    if smallest is None or n < smallest:
        smallest = n

print("Smallest:", smallest)
```

```python
numbers = {10, 15, 20, 25, 30, 35}

for n in numbers:
    if n % 2 == 0:
        print(n)
```
