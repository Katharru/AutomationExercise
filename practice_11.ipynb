```python
numbers = {"a": 10, "b": 20, "c": 30}
total = 0
count = 0

for value in numbers.values():
    total += value
    count += 1

if count > 0:
    print("Average:", total / count)
else:
    print("Dictionary is empty")
```

```python
numbers = {"a": 10, "b": -5, "c": 20, "d": -8}
positive = 0
negative = 0

for value in numbers.values():
    if value > 0:
        positive += 1
    elif value < 0:
        negative += 1

print("Positive:", positive)
print("Negative:", negative)
```

```python
marks = {"Rahul": 85, "Priya": 95, "Amit": 70}
topper = ""
highest = -1

for name, mark in marks.items():
    if mark > highest:
        highest = mark
        topper = name

print("Topper:", topper)
print("Marks:", highest)
```
