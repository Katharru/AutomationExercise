```python
r = float(input("Enter radius: "))
h = float(input("Enter height: "))

v = 3.14159 * r ** 2 * h
print("Volume of cylinder =", v)
```

```python
r = float(input("Enter radius: "))
h = float(input("Enter height: "))

v = (3.14159 * r ** 2 * h) / 3
print("Volume of cone =", v)
```
