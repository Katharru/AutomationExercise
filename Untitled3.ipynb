```python
text = input("Enter a string: ")
result = ""

for ch in text:
    if ch.lower() in "aeiou":
        result += "*"
    else:
        result += ch

print(result)
```

```python
text = input("Enter a string: ")
result = ""

for ch in text:
    if ch.isalnum() or ch == " ":
        result += ch

print(result)
```

```python
text = input("Enter a string: ").lower()

if text == text[::-1]:
    print("Palindrome")
else:
    print("Not a palindrome")
```

```python
text = input("Enter a string: ")

if text.isalpha():
    print("Only letters")
else:
    print("Contains other characters")
```
