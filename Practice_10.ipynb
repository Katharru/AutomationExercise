```python
student = {"name": "Rahul", "age": 10, "class": 5}

for key, value in student.items():
    print(key, ":", value)
```

```python
marks = {
    "Math": 90,
    "English": 85,
    "Science": 95
}

total = sum(marks.values())
print("Total Marks:", total)
```

```python
prices = {"Shirt": 500, "Shoes": 1500, "Bag": 800}

for item, price in prices.items():
    if price >= 1000:
        discount = price * 0.10
    else:
        discount = price * 0.05

    final_price = price - discount
    print(item, "Final price:", final_price)
```

```python
students = {
    1: "Amit",
    2: "Rahul",
    3: "Anita",
    4: "Sneha"
}

for roll, name in students.items():
    if name.startswith("A"):
        print(name)
    else:
        print("Name does not start with A:", name)
```

```python
marks = {"Rahul": 85, "Priya": 30, "Amit": 70}

for name, mark in marks.items():
    if mark >= 35:
        print(name, "Passed")
    else:
        print(name, "Failed")
```

```python
numbers = {"a": 25, "b": 60, "c": 80, "d": 40}
count = 0

for value in numbers.values():
    if value > 50:
        count += 1

print("Count:", count)
```

```python
numbers = {"a": 10, "b": 15, "c": 20, "d": 25}

for key, value in numbers.items():
    if value % 2 == 0:
        print(value, "is even")
    else:
        print(value, "is odd")
```

```python
numbers = {"a": 10, "b": -5, "c": 0}

for key, value in numbers.items():
    if value > 0:
        print(value, "Positive")
    elif value < 0:
        print(value, "Negative")
    else:
        print(value, "Zero")
```

```python
numbers = {"a": 10, "b": 20, "c": 30}
total = 0

for value in numbers.values():
    total += value

print("Sum:", total)
```
