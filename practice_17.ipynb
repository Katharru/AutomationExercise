```python
numbers = {10, 15, 20, 25, 30, 35}
even = 0
odd = 0

for n in numbers:
    if n % 2 == 0:
        even += 1
    else:
        odd += 1

print("Even:", even)
print("Odd:", odd)
```

```python
a = {1, 2, 3}
b = {3, 4, 5}
result = set()

for n in a:
    result.add(n)

for n in b:
    result.add(n)

print("Union:", result)
```
